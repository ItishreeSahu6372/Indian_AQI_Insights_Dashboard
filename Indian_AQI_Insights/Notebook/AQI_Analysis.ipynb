# India AQI Insights : A Data Analytics Study of Indian Cities (2022-2025)

## 1. Project Overview

Air pollution is one of the major environmental challenges affecting public health and quality of life.

This project analyzes Air Quality Index (AQI) data collected from multiple Indian cities between 2022 and 2025. The goal is to understand pollution trends, identify highly polluted cities, determine dominant pollutants, and evaluate overall air quality conditions.

## 2. Objectives

The objectives of this project are:

- Analyze AQI trends over time.
- Identify cities with the highest pollution levels.
- Determine the most frequently occurring pollutants.
- Examine the distribution of air quality categories.
- Prepare a clean dataset for dashboard development in Excel.

##  3. Dataset Overview
This dataset contains comprehensive records of air quality across various states and cities of India, including AQI values, pollutants, and air quality statuses over time.

Total Records: 235,785
Time Span: Multiple dates (latest: 30-April-2025)
- Coverage: 32 States and 291 cities
  
##### Features Explained
Column Name and	Description
- date :    Date of observation (DD-MM-YYYY format)
- state :	Indian state name
- area:	  City or locality where the reading was recorded
- number_of_monitoring_stations:	Number of stations in the area recording AQI
- prominent_pollutants: 	Dominant pollutant for that location on the given day (e.g., PM2.5, PM10, CO, NO2, O3)
- aqi_value:	Air Quality Index value (ranging from 0 to 500)
- air_quality_status:	AQI category (e.g., Good, Satisfactory, Moderate, Poor, Very Poor, Severe)
- unit: 	Description unit of AQI (textual, mostly constant)

## 6. Data Cleaning

```python
# keep required columns
df = df[
    [
        'date',
        'area',
        'prominent_pollutants',
        'aqi_value',
        'state'
    ]
]
```

```python
# check and drop missing values
missing = df.isnull().sum()
print(missing)
```

```python
# date conversion (the data become a date time object now)
df['date'] = pd.to_datetime(df['date'])
```

#### Data Cleaning Observations
- Selected only the relevant columns required for AQI analysis:
Date, state, City (Area), Prominent Pollutant, AQI Value
- Checked for missing values in all selected columns. No missing values were found in the dataset.
- Converted the date column from string format to datetime format for time-based analysis.

## 7. Feature Engineering

```python
# create Month-year (Extract only the month and year)
df['month_year'] = df['date'].dt.to_period('M')
```

```python
# create Year (Extract only the year)
df['year'] = df['date'].dt.year
```

```python
# Create Month Name
df['month'] = df['date'].dt.month_name()
```



#### Feature Engineering Observations

* Created a **Month-Year** column for monthly trend analysis.
* Extracted **Year** from the date column for yearly analysis.
* Extracted **Month** from the date column to identify monthly patterns.

## 8. Exploratory Data Analysis

#### 8.1. AQI Trend

```python
monthly_trend = (
    df.groupby('month_year')['aqi_value'].mean()
      .reset_index()
)
monthly_trend.head()
```

#### 8.2. Top Polluted Cities

```python
top_cities = (
    df.groupby('area')['aqi_value']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)
top_cities.head()
```

#### 8.3. Pollutant Analysis

```python
pollutants = (
    df['prominent_pollutants']
    .str.split(',')
    .explode()
    .str.strip()
)

pollutant_distribution = (
    pollutants
    .value_counts()
    .reset_index()
)

pollutant_distribution.columns = [
    'Pollutant',
    'Count'
]

pollutant_distribution.head()
```

####  8.4. monthly (state & city-wise) AQI Analysis

```python
monthly_aqi = (
    df.groupby(['month_year', 'state', 'area'])['aqi_value']
      .mean()
      .reset_index()
)

monthly_aqi.rename(columns={
    'area': 'City',
    'aqi_value': 'Avg_AQI',
    'state': 'State'
}, inplace=True)

monthly_aqi.head()
```

#### 8.5. State-wise AQI analysis

```python
state_aqi = (
    df.groupby('state')['aqi_value']
      .mean()
      .reset_index()
)

state_aqi.rename(columns={
    'state': 'State',
    'aqi_value': 'Avg_AQI'
}, inplace=True)

state_aqi.sort_values(by='Avg_AQI',ascending=False,inplace=True)
state_aqi.head()
```

#### Key Insights

* Air quality levels showed noticeable variation across different months, indicating seasonal changes in pollution patterns.
* Byrnihat, Begusarai, and Delhi ranked among the most polluted cities based on average AQI values.
* PM10 and PM2.5 were the most frequently reported pollutants, highlighting particulate matter as a major contributor to air pollution.
* Significant differences in average AQI were observed across Indian states, indicating regional variations in air quality conditions.

## 9. Export for Excel Dashboard

```python
with pd.ExcelWriter(r"D:\project\my project dataset\AQI_Dashboard_Data.xlsx") as writer:

    monthly_aqi.to_excel(
        writer,
        sheet_name="Monthly_AQI",
        index=False
    )

    state_aqi.to_excel(
        writer,
        sheet_name="State_AQI",
        index=False
    )

    pollutant_distribution.to_excel(
        writer,
        sheet_name="Pollutants",
        index=False
    )
```
